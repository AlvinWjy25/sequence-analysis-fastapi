```python
def inspect_fasta(file_path):
    with open(file_path, "r") as f:
        lines = [line.strip() for line in f.readlines() if line.strip()]

    # 1. Cek Record & Header
    header_lines = [line for line in lines if line.startswith(">")]
    if len(header_lines) != 1:
        return f"Error: found {len(header_lines)} record (have to be exactly 1 record)"

    header = header_lines[0]
    sequence_id = header[1:].split()[0]  #First token after '>'

    # 2. Merge Sequence and Capitalize
    seq_lines = [line for line in lines if not line.startswith(">")]
    raw_sequence = "".join(seq_lines).replace(" ", "").upper()

    # 3. Validate Characters
    valid_bases = set("ACGTN")
    invalid_chars = set(raw_sequence) - valid_bases
    if invalid_chars:
        return f"Error: Invalid Character detected: {invalid_chars}"

    # 4. Validate the length of sequence
    seq_len = len(raw_sequence)
    if not (1 <= seq_len <= 1000):
        return f"Error: Length of Sequence {seq_len} bp (must be between 1-1000 bp)"

    return {
        "sequence_id": sequence_id,
        "length": seq_len,
        "n_count": raw_sequence.count("N"),
        "sequence_preview": (
            raw_sequence[:30] + "..." if seq_len > 30 else raw_sequence
        ),
    }


# Uji coba dengan file patient
print(inspect_fasta(SUPPORTING_DATA / 'patient_01.fasta'))
print(inspect_fasta(SUPPORTING_DATA / 'patient_02.fasta'))
print(inspect_fasta(SUPPORTING_DATA / 'patient_03.fasta'))
print(inspect_fasta(SUPPORTING_DATA / 'patient_04.fasta'))
print(inspect_fasta(SUPPORTING_DATA / 'patient_05.fasta'))

print(inspect_fasta(SUPPORTING_DATA / 'reference.fasta'))
```
